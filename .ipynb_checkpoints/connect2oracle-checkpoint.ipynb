```python
import oracledb
import pandas as pd
import numpy as np
import os
# !pip install python-dotenv
from dotenv import load_dotenv

import seaborn as sns
import matplotlib.pyplot as plt

# !pip install yellowbrick
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from yellowbrick.cluster import KElbowVisualizer
from sklearn.metrics import silhouette_score
from yellowbrick.cluster import SilhouetteVisualizer
```

```python
# 1. .env 파일로부터 환경 변수 로드
load_dotenv()

# 2. 환경 변수에서 DB 접속 정보 가져오기 (코드에 노출되지 않음)
DB_USER = os.getenv("DB_USER")
DB_PASSWORD = os.getenv("DB_PASSWORD")
DB_HOST = os.getenv("DB_HOST")
DB_PORT = os.getenv("DB_PORT")
DB_SERVICE_NAME = os.getenv("DB_SERVICE_NAME")

# 이메일이나 주소창처럼 하나의 문자열(DSN)로 결합
dsn_string = f"{DB_HOST}:{DB_PORT}/{DB_SERVICE_NAME}"
```

```python
rfm_query = """with Payment_view as 
(select order_id, sum(payment_value) Monetary
from olist_order_payments_dataset
group by order_id
)
select customer_unique_id, TRUNC(TO_DATE('2018-08-30','YYYY-MM-DD')) -TRUNC(MAX(order_purchase_timestamp)) as Recency, 
count(C.customer_unique_id) as Frequency, 
sum(P.Monetary) as Monetary
from olist_orders_dataset O, olist_customers_dataset C, Payment_view P
where O.order_id = P.order_id 
and O.customer_id = C.customer_id
and O.order_status='delivered'
and O.order_purchase_timestamp >= TO_DATE('2017-01-01','YYYY-MM-DD')
and O.order_purchase_timestamp < TO_DATE('2018-09-01','YYYY-MM-DD')
group by customer_unique_id
order by Recency desc, frequency desc, monetary desc"""
```

```python
connection = None
try:
    # 3. DB 연결 수행 (Thin 모드로 다이렉트 접속)
    connection = oracledb.connect(
        user=DB_USER,
        password=DB_PASSWORD,
        dsn=dsn_string
    )
    print(" 오라클 데이터베이스 연결 성공!")

    # 4. Pandas를 이용해 SQL 결과를 데이터프레임으로 바로 읽어오기
    df_rfm = pd.read_sql_query(rfm_query, con=connection)
    print(f" 데이터 로드 완료! (총 {len(df_rfm)}행 조회됨)")
    
    # 데이터 상위 5줄 출력 확인
    print(df_rfm.head())

except Exception as e:
    print(f" DB 연동 중 에러 발생: {e}")

finally:
    # 5. [중요] 사용이 끝난 DB 연결은 반드시 닫아서 세션을 반환합니다.
    if connection:
        connection.close()
        print(" 데이터베이스 연결 닫힘.")
```

```python
#데이터 정규화 
#로그변환후(이상치)
rfm_log = np.log1p(df_rfm)
StandardScaler = StandardScaler()
df_scale = StandardScaler.fit_transform(rfm_log)

df_scale = pd.DataFrame(data = df_scale, columns = df_rfm.columns)
# df_scale['Customer_unique_id'] = df_rfm_unique_id
df_scale.head()
```

```python
#kmeans 클러스터링모델 만들기
kmeans_model = KMeans(
    init = 'k-means++',
    n_clusters = 4,
    n_init = 10,
    max_iter = 300
)

kmeans_model.fit(df_scale)

print(kmeans_model.inertia_)

print(kmeans_model.cluster_centers_)

print(kmeans_model.n_iter_)

##엘보우차트 시각화
Elbow_chart = KElbowVisualizer(kmeans_model,k=(1,11))
Elbow_chart.fit(df_scale)
Elbow_chart.draw()
```

```python
# 1. 최적의 K=4로 최종 모델 학습 및 라벨링
final_kmeans = KMeans(n_clusters=4, init='k-means++', random_state=42, n_init=10)
df_rfm['CLUSTER'] = final_kmeans.fit_predict(df_scale)

# 2. 군집별로 실제 RFM 원본 값의 평균과 고객 수 계산 (프로파일링)
cluster_profile = df_rfm.groupby('CLUSTER').agg({
    'RECENCY': ['mean','std'],
    'FREQUENCY': 'mean',
    'MONETARY': ['mean','std'],
    'CLUSTER': 'count'
}).rename(columns={'CLUSTER': 'CUSTOMER_COUNT'}).reset_index()

print("\n [최종 군집별 RFM 프로파일링 결과]")
print(cluster_profile)
```

```python
final_kmeans = KMeans(n_clusters=4, init='k-means++', random_state=42, n_init=10)
df_scale['cluster'] = final_kmeans.fit_predict(df_scale)
df_scale.head()
```

```python
# 1. 클러스터별 금액(sum) 비중과 고객수(count) 비중을 동시에 계산
df_group = df_rfm.groupby('CLUSTER')['MONETARY'].agg(['sum', 'count']).reset_index()

# 전체 합으로 나누어 각각의 비중(Ratio) 구하기
df_group['MONETARY_RATIO'] = df_group['sum'] / df_rfm['MONETARY'].sum()
df_group['COUNT_RATIO'] = df_group['count'] / df_rfm['MONETARY'].count()
```
